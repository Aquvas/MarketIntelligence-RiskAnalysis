```python
from pathlib import Path

# Work on a copy so the original downloaded data stays unchanged
df_clean = df.copy()

# Convert the date index into a regular column
df_clean = df_clean.reset_index()
df_clean = df_clean.rename(columns={"Datetime": "date", "Date": "date"})

# Add the ticker identifier
df_clean["ticker"] = "NVDA"

# Standardize column names
df_clean.columns = [
    col.strip().lower().replace(" ", "_")
    for col in df_clean.columns
]

# Arrange columns consistently
df_clean = df_clean[
    [
        "date",
        "ticker",
        "open",
        "high",
        "low",
        "close",
        "adj_close",
        "volume",
        "dividends",
        "stock_splits",
    ]
]

print(df_clean.head())
print(df_clean.columns.tolist())
print("Shape:", df_clean.shape)
```

```python
data_dir = Path("C:\my Files\RiskPredANDMarketEvaluation\data\processed")
data_dir.mkdir(parents=True, exist_ok=True)

output_path = data_dir / "market_data.csv"

df_clean.to_csv(output_path, index=False)

print("Saved to:", output_path.resolve())
```
